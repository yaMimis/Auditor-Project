```python
import sys; sys.path.append('../..')
from src.modelos.dados import ler_base, preparar, dividir
from src.modelos.variaveis import criar_variaveis, colunas_do_modelo
from src.modelos.treino import (montar_baseline, validar_cruzado, treinar_e_avaliar,
                                comparar_com_bondora, salvar)

print('Variáveis no modelo:', len(colunas_do_modelo()))   # tem que dar 30

base = preparar(ler_base())
treino, teste = dividir(base)
treino = criar_variaveis(treino)
teste = criar_variaveis(teste, referencia=treino)

cv = validar_cruzado(montar_baseline(), treino)
print(cv)
print('Média:', cv[['auc', 'ks', 'brier']].mean().round(4).to_dict())

r = treinar_e_avaliar(montar_baseline(), treino, teste)
print('\nTeste:', r['teste'])
print('\nPor país:\n', r['teste_por_pais'])
print('\nCalibração:\n', r['calibracao'])
print('\n', comparar_com_bondora(teste, r['p_teste']))

salvar(r['modelo'], 'baseline_logistica_v2',
       {'cv_media': cv[['auc', 'ks', 'brier']].mean().round(4).to_dict(), 'teste': r['teste']})
```
